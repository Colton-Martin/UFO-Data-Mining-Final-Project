## Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules

from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score
from sklearn.neighbors import NearestNeighbors

RANDOM_STATE = 42
```

## Load NUFROC Dataset

```python
ufo = pd.read_csv('Data-Sources/Original-Data-NotCleaned/nuforc_reports.csv')

print("Original entries:", len(ufo))
print("Original columns:", len(ufo.columns))
```

## Standardize Column Names

```python
# Standardize column names
ufo.columns = [
    c.strip().lower().replace(" ", "_")
    for c in ufo.columns
]

ufo.columns
```

## Convert Date Time and Create Year

```python
# Convert date_time to datetime
ufo['date_time'] = pd.to_datetime(
    ufo['date_time'],
    errors='coerce'
)

# Extract year
ufo['year'] = ufo['date_time'].dt.year
```

## Define Valid US States & Filter out Non-States

```python
valid_states = [
    'AL','AK','AZ','AR','CA','CO','CT','DE','FL','GA',
    'HI','ID','IL','IN','IA','KS','KY','LA','ME','MD',
    'MA','MI','MN','MS','MO','MT','NE','NV','NH','NJ',
    'NM','NY','NC','ND','OH','OK','OR','PA','RI','SC',
    'SD','TN','TX','UT','VT','VA','WA','WV','WI','WY','DC'
]

# Identify non-U.S. or invalid state values
filtered_states = ufo[
    ~ufo['state'].isin(valid_states)
]['state'].value_counts(dropna=False)

filtered_states
```

## Identify to US Sightings, 2010-2019

```python
# Filter to U.S. states and years 2010–2019
ufo_2010_2019 = ufo[
    (ufo['state'].isin(valid_states)) &
    (ufo['year'] >= 2010) &
    (ufo['year'] <= 2019)
].copy()

print("Entries after U.S. and year filtering:", len(ufo_2010_2019))

ufo_2010_2019['year'].value_counts().sort_index()
```

## Handle Missing Values

```python
# Missing values by column
ufo_2010_2019.isnull().sum()

missingness = pd.DataFrame({
    'Missing Values': ufo_2010_2019.isnull().sum(),
    'Missing Percentage': (
        ufo_2010_2019.isnull().mean() * 100
    )
})

missingness
```

## Filter out Missing City Values

```python
# Number of entries before removing missing city/latitude/longitude values
before = len(ufo_2010_2019)

# Remove rows where city, city latitude, or city longitude is null
ufo_2010_2019 = ufo_2010_2019.dropna(
    subset=['city', 'city_latitude', 'city_longitude']
).copy()

# Number of entries after removal
after = len(ufo_2010_2019)

# Calculate number and percentage removed
rows_removed = before - after
percent_removed = (rows_removed / before) * 100

print("Entries before:", before)
print("Entries removed:", rows_removed)
print("Entries after:", after)
print("Percent of rows removed:", round(percent_removed, 2), "%")

# Confirm no missing values remain
print("\nMissing city values:", ufo_2010_2019['city'].isnull().sum())
print("Missing latitude values:", ufo_2010_2019['city_latitude'].isnull().sum())
print("Missing longitude values:", ufo_2010_2019['city_longitude'].isnull().sum())
```

## Extract Month, Day, and Hour, Create Season, and Create day/night Variable

```python
# Create a copy for additional preprocessing
ufo_clean = ufo_2010_2019.copy()

# Extract date/time components
ufo_clean['month'] = ufo_clean['date_time'].dt.month
ufo_clean['day'] = ufo_clean['date_time'].dt.day
ufo_clean['hour'] = ufo_clean['date_time'].dt.hour

def get_season(month):
    if month in [12, 1, 2]:
        return 'winter'
    elif month in [3, 4, 5]:
        return 'spring'
    elif month in [6, 7, 8]:
        return 'summer'
    else:
        return 'fall'

ufo_clean['season'] = ufo_clean['month'].apply(get_season)

ufo_clean['season'].value_counts()

# 1 = nighttime, 0 = daytime
ufo_clean['is_night'] = ufo_clean['hour'].apply(
    lambda h: 1 if (h >= 18 or h < 6) else 0
)
```

## Group Uncommon Shapes

```python
common_shapes = [
    'light',
    'circle',
    'triangle',
    'disk',
    'sphere',
    'fireball'
]

ufo_clean['shape_clean'] = np.where(
    ufo_clean['shape'].isin(common_shapes),
    ufo_clean['shape'],
    'unknown'
)

ufo_clean['shape_clean'].value_counts()
```

## Extract Duration in Seconds

```python
ufo_clean['duration_sec'] = (
    ufo_clean['duration']
    .str.extract(
        r'(\d+\.?\d*)',
        expand=False
    )
    .astype(float)
)

ufo_clean['duration_log'] = np.log1p(
    ufo_clean['duration_sec']
)
```

```python
print("Final entries:", len(ufo_clean))
print("Final columns:", len(ufo_clean.columns))
ufo_clean.head()
```

### Geofence Reaffirmation Using Coords

```python
# Remove sightings south of Hawaii's southernmost point
# and east of the easternmost point of the 50 U.S. states

ufo_clean = ufo_clean[
    (ufo_clean['city_latitude'] >= 18.9) &
    (ufo_clean['city_longitude'] <= -66.9)
].copy()

print("Remaining UFO sightings:", len(ufo_clean))
print("Minimum latitude:", ufo_clean['city_latitude'].min())
print("Maximum longitude:", ufo_clean['city_longitude'].max())

plt.figure(figsize=(14, 8))

plt.scatter(
    ufo_clean['city_longitude'],
    ufo_clean['city_latitude'],
    s=2,
    alpha=0.3
)

plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('UFO Sightings After Non-US Removal')

plt.show()
```

## Clustering Analysis
In this section, we apply k-means clustering and DBSCAN to the latitude and longitude of the cities (city_latitude and city_longitude) for the previously cleaned dataset. This produces clusters which represent where most reports have been received from. It does not represent genuine UFO behaviour.

## Prepare Geographical Features

```python
# Keep the geographic coordinates needed for clustering
cluster_data = ufo_clean[
    ['city_latitude', 'city_longitude']
].copy()

# Confirm that the coordinates are numeric and valid
cluster_data['city_latitude'] = pd.to_numeric(
    cluster_data['city_latitude'], errors='coerce'
)
cluster_data['city_longitude'] = pd.to_numeric(
    cluster_data['city_longitude'], errors='coerce'
)

cluster_data = cluster_data.dropna().copy()

# Remove coordinates outside valid latitude/longitude ranges
cluster_data = cluster_data[
    cluster_data['city_latitude'].between(-90, 90) &
    cluster_data['city_longitude'].between(-180, 180)
].copy()

print("Rows available for clustering:", len(cluster_data))
print(cluster_data[['city_latitude', 'city_longitude']].describe())

# Convert latitude and longitude to approximate kilometers.
# This uses a simple equirectangular projection centered on the mean latitude.
earth_radius_km = 6371.0
mean_latitude = np.radians(cluster_data['city_latitude'].mean())

cluster_data['x_km'] = (
    earth_radius_km *
    np.radians(cluster_data['city_longitude']) *
    np.cos(mean_latitude)
)
cluster_data['y_km'] = (
    earth_radius_km *
    np.radians(cluster_data['city_latitude'])
)

X = cluster_data[['x_km', 'y_km']].copy()

# Keep the original dataframe index so cluster labels can be joined back.
print("Geographic feature matrix shape:", X.shape)
```

## K-Means Clustering

```python
# Test several values of K and compare inertia and silhouette score
k_values = range(2, 9)
kmeans_results = []

for k in k_values:
    model = KMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        n_init=10
    )
    labels = model.fit_predict(X)

    # Use a sample for silhouette scoring if the dataset is large
    sample_size = min(10000, len(X))
    score = silhouette_score(
        X,
        labels,
        sample_size=sample_size,
        random_state=RANDOM_STATE
    )

    kmeans_results.append({
        'k': k,
        'inertia': model.inertia_,
        'silhouette_score': score
    })

kmeans_results = pd.DataFrame(kmeans_results)
display(kmeans_results)
```

```python
# Elbow method: look for a point where inertia begins to decrease more slowly
plt.figure(figsize=(8, 5))
plt.plot(kmeans_results['k'], kmeans_results['inertia'], marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('K-Means Elbow Method')
plt.xticks(list(k_values))
plt.tight_layout()
plt.show()
```

```python
# Compare silhouette scores across K values
plt.figure(figsize=(8, 5))
plt.plot(
    kmeans_results['k'],
    kmeans_results['silhouette_score'],
    marker='o'
)
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.title('K-Means Silhouette Scores')
plt.xticks(list(k_values))
plt.tight_layout()
plt.show()
```

```python
# Set this after reviewing the elbow and silhouette plots.
# Change the value if another K is more appropriate for the data.
BEST_K = int(
    kmeans_results.loc[
        kmeans_results['silhouette_score'].idxmax(), 'k'
    ]
)

kmeans_model = KMeans(
    n_clusters=BEST_K,
    random_state=RANDOM_STATE,
    n_init=10
)

cluster_data['kmeans_cluster'] = kmeans_model.fit_predict(X)

print("Selected K:", BEST_K)
print("K-Means silhouette score:",
      round(kmeans_results.loc[
          kmeans_results['k'] == BEST_K,
          'silhouette_score'
      ].iloc[0], 4))
```

```python
# Plot K-Means clusters using longitude and latitude
plt.figure(figsize=(12, 7))
scatter = plt.scatter(
    cluster_data['city_longitude'],
    cluster_data['city_latitude'],
    c=cluster_data['kmeans_cluster'],
    cmap='tab10',
    s=5,
    alpha=0.45
)
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title(f'K-Means Geographic Clusters (K={BEST_K})')
plt.colorbar(scatter, label='Cluster')
plt.tight_layout()
plt.show()
```

```python
# Summarize the size and geographic center of each K-Means cluster
kmeans_summary = cluster_data.groupby('kmeans_cluster').agg(
    sightings=('city_latitude', 'size'),
    average_latitude=('city_latitude', 'mean'),
    average_longitude=('city_longitude', 'mean')
).reset_index()

kmeans_summary['percentage'] = (
    kmeans_summary['sightings'] / len(cluster_data) * 100
)

display(kmeans_summary.sort_values('sightings', ascending=False))
```

## DBSCAN Clustering

```python
# Use the 10th-nearest-neighbor distance plot to help select DBSCAN epsilon.
# Distances are in the approximate kilometer feature space created above.
MIN_SAMPLES = 10

neighbors = NearestNeighbors(n_neighbors=MIN_SAMPLES)
neighbors_fit = neighbors.fit(X)
distances, indices = neighbors_fit.kneighbors(X)

k_distances = np.sort(distances[:, MIN_SAMPLES - 1])

plt.figure(figsize=(8, 5))
plt.plot(k_distances)
plt.xlabel('Points sorted by distance')
plt.ylabel(f'Distance to {MIN_SAMPLES}th nearest neighbor (km)')
plt.title('DBSCAN k-Distance Plot')
plt.tight_layout()
plt.show()
```

```python
# Test several epsilon values. Use the k-distance plot and these results
# together when choosing the final epsilon.
eps_values = [10, 25, 50, 75, 100]
dbscan_tests = []

for eps in eps_values:
    model = DBSCAN(
        eps=eps,
        min_samples=MIN_SAMPLES,
        metric='euclidean'
    )
    labels = model.fit_predict(X)

    cluster_count = len(set(labels) - {-1})
    noise_count = int((labels == -1).sum())

    dbscan_tests.append({
        'eps_km': eps,
        'clusters_found': cluster_count,
        'noise_points': noise_count,
        'noise_percentage': noise_count / len(labels) * 100
    })

dbscan_parameter_results = pd.DataFrame(dbscan_tests)
display(dbscan_parameter_results)
```

```python
# Update EPS_KM after reviewing the plots and parameter-test table.
EPS_KM = 50

dbscan_model = DBSCAN(
    eps=EPS_KM,
    min_samples=MIN_SAMPLES,
    metric='euclidean'
)

cluster_data['dbscan_cluster'] = dbscan_model.fit_predict(X)

dbscan_cluster_count = len(
    set(cluster_data['dbscan_cluster']) - {-1}
)
dbscan_noise_count = int(
    (cluster_data['dbscan_cluster'] == -1).sum()
)
dbscan_noise_percentage = (
    dbscan_noise_count / len(cluster_data) * 100
)

print("DBSCAN epsilon (km):", EPS_KM)
print("DBSCAN clusters:", dbscan_cluster_count)
print("Noise points:", dbscan_noise_count)
print("Noise percentage:", round(dbscan_noise_percentage, 2), "%")
```

```python
# Plot DBSCAN clusters. Label -1 represents noise points.
plt.figure(figsize=(12, 7))
scatter = plt.scatter(
    cluster_data['city_longitude'],
    cluster_data['city_latitude'],
    c=cluster_data['dbscan_cluster'],
    cmap='tab20',
    s=5,
    alpha=0.45
)
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title(f'DBSCAN Geographic Clusters (eps={EPS_KM} km)')
plt.colorbar(scatter, label='Cluster (-1 = noise)')
plt.tight_layout()
plt.show()
```

```python
# Summarize DBSCAN cluster sizes and geographic centers.
# Noise points (-1) are shown separately in this summary.
dbscan_summary = cluster_data.groupby('dbscan_cluster').agg(
    sightings=('city_latitude', 'size'),
    average_latitude=('city_latitude', 'mean'),
    average_longitude=('city_longitude', 'mean')
).reset_index()

dbscan_summary['percentage'] = (
    dbscan_summary['sightings'] / len(cluster_data) * 100
)

display(dbscan_summary.sort_values('sightings', ascending=False))
```

## Compare K-Means and DBSCAN

```python
# Compare the number of clusters and the number of points assigned.
comparison = pd.DataFrame({
    'Method': ['K-Means', 'DBSCAN'],
    'Clusters': [
        cluster_data['kmeans_cluster'].nunique(),
        dbscan_cluster_count
    ],
    'Points labeled as noise': [
        0,
        dbscan_noise_count
    ],
    'Noise percentage': [
        0,
        dbscan_noise_percentage
    ]
})

display(comparison)
```

```python
# Calculate silhouette scores.
# DBSCAN's score is calculated only on non-noise observations, so it is
# not directly equivalent to the K-Means score.
kmeans_silhouette = silhouette_score(
    X,
    cluster_data['kmeans_cluster'],
    sample_size=min(10000, len(X)),
    random_state=RANDOM_STATE
)

dbscan_non_noise = cluster_data['dbscan_cluster'] != -1
dbscan_labels_non_noise = cluster_data.loc[
    dbscan_non_noise, 'dbscan_cluster'
]

if dbscan_labels_non_noise.nunique() >= 2:
    dbscan_silhouette = silhouette_score(
        X.loc[dbscan_non_noise],
        dbscan_labels_non_noise,
        sample_size=min(10000, dbscan_non_noise.sum()),
        random_state=RANDOM_STATE
    )
else:
    dbscan_silhouette = np.nan

print("K-Means silhouette score:",
      round(kmeans_silhouette, 4))
print("DBSCAN silhouette score excluding noise:",
      dbscan_silhouette)
```

## Prepare Categorical Features

```python
# Create duration and time categories for association-rule mining.
ufo_clean['duration_sec'] = (
    ufo_clean['duration']
    .astype(str)
    .str.extract(r'(\d+\.?\d*)', expand=False)
    .astype(float)
)

ufo_clean['duration_bin'] = pd.cut(
    ufo_clean['duration_sec'],
    bins=[0, 60, 300, np.inf],
    labels=['short', 'medium', 'long']
)

ufo_clean['time_bin'] = pd.cut(
    ufo_clean['hour'],
    bins=[-1, 5, 11, 17, 23],
    labels=['early_morning', 'morning', 'afternoon', 'evening']
)

categorical_cols = [
    'shape_clean',
    'season',
    'duration_bin',
    'time_bin'
]

association_data = ufo_clean[categorical_cols].copy()
association_data = association_data.dropna()

ufo_dummies = pd.get_dummies(
    association_data,
    prefix=categorical_cols
)

ufo_dummies.head()
```

## Apriori Frequent Itemsets and Association Rules

```python
freq_items_apriori = apriori(
    ufo_dummies.astype(bool),
    min_support=0.05,
    use_colnames=True
)

display(
    freq_items_apriori.sort_values(
        'support', ascending=False
    ).head(20)
)
```

```python
rules = association_rules(
    freq_items_apriori,
    metric='lift',
    min_threshold=1.0
)

rules_sorted = rules.sort_values(
    'lift', ascending=False
)

display(rules_sorted.head(20))
```

## FP-Growth Frequent Itemsets

```python
freq_items_fpgrowth = fpgrowth(
    ufo_dummies.astype(bool),
    min_support=0.05,
    use_colnames=True
)

display(
    freq_items_fpgrowth.sort_values(
        'support', ascending=False
    ).head(20)
)
```
